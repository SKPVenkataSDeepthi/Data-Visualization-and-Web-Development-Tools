A Box Plot is also known as a Whisker Plot and is a standardized way of displaying the distribution of data based on a five-number summary: minimum, first quartile (Q1), median (Q2), third quartile (Q3) and maximum. It can also show outliers

```python
import matplotlib.pyplot as plt
import pandas as pd

# Load the custom dataset (replace the path with the actual path to your CSV file)
df = pd.read_csv('Airline Dataset Updated - v2.csv')

df.head()

# Extract the Age column for the box plot
ages = df['Age'].values

# Create a box plot
plt.boxplot(ages, vert=True, patch_artist=True, 
            boxprops=dict(facecolor='skyblue'),
            medianprops=dict(color='red'))

plt.xlabel('Age')
plt.ylabel('Values')
plt.title('Box Plot of Age')
plt.show()
```

```python
# 2. Horizontal Box Plot
plt.boxplot(df['Age'], vert=False, patch_artist=True, 
            boxprops=dict(facecolor='lightgreen'),
            medianprops=dict(color='purple'))
```

```python
# 3. Customizing whiskers and fillers
plt.boxplot(df['Age'], vert=True, patch_artist=True, 
            whiskerprops=dict(color='orange', linewidth=2),
            flierprops=dict(marker='o', color='purple', markersize=8),
            boxprops=dict(facecolor='lightblue'),
            medianprops=dict(color='red'))
```

```python
# 4. Adding notches --- represent the confidence interval around the median
plt.boxplot(df['Age'], vert=True, patch_artist=True, notch=True,
            boxprops=dict(facecolor='lightyellow'),
            medianprops=dict(color='black'))
```

```python
# 5. changing box plot width
plt.boxplot(df['Age'], vert=True, patch_artist=True, widths=0.5, 
            boxprops=dict(facecolor='lightblue'),
            medianprops=dict(color='red'))
```

```python
# 6. Adding grid
plt.boxplot(df['Age'], vert=True, patch_artist=True, 
            boxprops=dict(facecolor='lightblue'),
            medianprops=dict(color='red'))
plt.grid(True)
```
